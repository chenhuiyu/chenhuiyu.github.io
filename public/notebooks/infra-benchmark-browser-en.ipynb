## 1 · Account for KV memory

Count both K and V; heads means KV heads. This is an estimate, not a measurement of GPU memory.

```python
# The blog provides live charts; standalone Python prints chart data.
if 'display_plot' not in globals():
    def display_plot(x, y, title='', xlabel='', ylabel=''):
        print(title, list(zip(x,y)))

layers, kv_heads, head_dim = 32, 8, 128
bytes_per_element, concurrency = 2, 8
def kv_gib(tokens):
    return 2*layers*kv_heads*head_dim*bytes_per_element*concurrency*tokens/2**30
lengths = [512,1024,2048,4096,8192]
for n in lengths:
    print(n, "tokens =>", kv_gib(n), "GiB")
display_plot(lengths, [kv_gib(n) for n in lengths], "KV estimate", "tokens / request", "GiB")
```

## 2 · Does tensor sharding preserve results?

This splits the input dimension, requiring a sum of partial products. Output-dimension sharding concatenates results instead.

```python
x = [2,3,5,7]
W = [[1,2],[3,4],[5,6],[7,8]]
full = [sum(x[i]*W[i][j] for i in range(4)) for j in range(2)]
parts = [[sum(x[i]*W[i][j] for i in shard) for j in range(2)] for shard in ([0,1],[2,3])]
merged = [sum(p[j] for p in parts) for j in range(2)]
print("local contributions:",parts)
print("all-reduce sum:",merged, "reference:",full)
assert merged == full
```

## 3 · ZeRO training states

16 bytes/parameter is this mixed-precision Adam convention: 2 weights, 2 gradients, 4 master weights, 8 moments. Activations are excluded.

```python
P, ranks = 7_000_000_000, 8
for stage in range(4):
    weights = 2*P/(ranks if stage>=3 else 1)
    gradients = 2*P/(ranks if stage>=2 else 1)
    optimizer = 12*P/(ranks if stage>=1 else 1)
    print("ZeRO",stage,"persistent GiB/rank:",(weights+gradients+optimizer)/2**30)
print("Peak also includes activations, temporary gathered parameters and buffers.")
```

## 4 · Actual browser CPU timing

This measures Python loops, not CUDA, BLAS or TPU. Fix inputs, warm up, then repeat. Change n to observe growth; do not compare serving frameworks with these numbers.

```python
import time, statistics
n = 24
matrix = [[(i+j)%7/7 for j in range(n)] for i in range(n)]
def matmul():
    return [[sum(matrix[i][k]*matrix[k][j] for k in range(n)) for j in range(n)] for i in range(n)]
matmul()
times = []
for _ in range(12):
    start = time.perf_counter()
    out = matmul()
    times.append((time.perf_counter()-start)*1000)
print("shape:",(n,n),"checksum:",sum(map(sum,out)))
print("median ms:",statistics.median(times),"max ms:",max(times))
display_plot(list(range(1,13)),times,"Measured Python matmul","repeat","ms")
```

## 5 · Where does the cached mask start?

Four cached tokens and two new tokens: predict the visible keys before running. Edit past_length/new_length. This constructs absolute-position visibility without padding.

```python
past_length, new_length = 4, 2
assert past_length >= 0 and new_length > 0
allowed = [[int(k <= past_length+q) for k in range(past_length+new_length)] for q in range(new_length)]
wrong = [[int(k <= q) for k in range(past_length+new_length)] for q in range(new_length)]
for q, (correct_row, wrong_row) in enumerate(zip(allowed, wrong)):
    print("absolute query",past_length+q,"correct:",correct_row,"unshifted:",wrong_row)
assert all(sum(row) == past_length+q+1 for q,row in enumerate(allowed))
display_plot(list(range(past_length,past_length+new_length)),[sum(row) for row in allowed],"Visible keys per new query","absolute query position","visible keys")
```

## 6 · Reuse K/V and compare with recomputation

Standard-library single-head attention with fixed scalar projections and stable softmax. Token-by-token outputs must match full causal computation. This isolates attention; the other tab contains the two-layer decoder, position embeddings, FP32/FP64 checks and CPU timings.

```python
import math
vectors = [[math.sin(i+j*.7) for j in range(4)] for i in range(8)]
def project(vector, scale):
    return [scale*x for x in vector]
def read_query(q, keys, values):
    scores = [sum(x*y for x,y in zip(q,k))/len(q)**.5 for k in keys]
    unnormalized = [math.exp(s-max(scores)) for s in scores]
    weights = [s/sum(unnormalized) for s in unnormalized]
    return [sum(a*v[d] for a,v in zip(weights,values)) for d in range(len(q))]
full = []
for i,x in enumerate(vectors):
    full.append(read_query(project(x,.7),[project(v,1.1) for v in vectors[:i+1]],[project(v,.9) for v in vectors[:i+1]]))
keys, values, cached = [], [], []
for x in vectors:
    keys.append(project(x,1.1))
    values.append(project(x,.9))
    cached.append(read_query(project(x,.7),keys,values))
errors = [max(abs(a-b) for a,b in zip(expected,actual)) for expected,actual in zip(full,cached)]
assert max(errors) < 1e-12
wrong_output = read_query(project(vectors[-1],.7),keys[:1],values[:1])
wrong_error = max(abs(a-b) for a,b in zip(full[-1],wrong_output))
assert wrong_error > 1e-3
print("cached K/V shape:",(len(keys),len(keys[0])))
print("max correct error:",max(errors),"unshifted-mask error:",wrong_error)
print("At query length 1, a top-left triangle can expose only key 0.")
display_plot(list(range(1,len(vectors)+1)),errors,"Full vs cached attention error","consumed tokens","max absolute error")
```
